# In this notebook
I make a lot of the small figures used in the summary figure.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("../data/durham_stations.csv")
```

```python
plt.figure(figsize=(1, 1))
plt.scatter(df['longitude'], df['latitude'], s=0.1)
plt.xticks([])
plt.yticks([])
plt.xlim(-79.1, -78.62)
plt.ylim(35.79, 36.2)
# Save fig
plt.savefig("../final_fig/station_map.pdf", dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(1, 1))
x = np.linspace(0, 10, 100)
y = x + np.random.normal(0, 1, size=x.shape)

ax.scatter(x, y, s=1)
ax.plot([0, 10], [0, 10], color='red')
ax.set_xticks([])
ax.set_yticks([])

plt.savefig("../final_fig/era5_wu_scatter.pdf", dpi=1000, bbox_inches='tight')
```

```python
x = np.linspace(0, 3, 100)
y = np.exp(-x)
fig, ax = plt.subplots(figsize=(1, 1))
ax.plot(x, y)
ax.set_xticks([])
ax.set_yticks([])
plt.savefig("../final_fig/matern_cov.pdf", dpi=300, bbox_inches='tight')
```

```python
x = np.linspace(0, 3, 100)
y = (1 + np.sqrt(3)*x**2)*np.exp(-np.sqrt(3)*x**2)
fig, ax = plt.subplots(figsize=(1, 1))
ax.plot(x, y)
ax.set_xticks([])
ax.set_yticks([])
plt.savefig("../final_fig/matern32_cov.pdf", dpi=300, bbox_inches='tight')
```

```python

fig, (ax, ax1) = plt.subplots(1, 2, figsize=(2, 1))


x = np.linspace(0, 10, 100)
y = x + np.random.normal(0, 1, size=x.shape)
ax.scatter(x, y, s=1)
ax.plot([0, 10], [0, 10], color='red')

ax.set_xticks([])
ax.set_yticks([])
ax.set_xlabel("ERA5")
ax.set_ylabel("WU")

y = 0.1*x + 0.5*np.random.normal(0, 1, size=x.shape)
ax1.scatter(x, y, s=1)
ax1.plot([0, 10], [0, 1], color='red')

ax1.set_xticks([])
ax1.set_yticks([])
ax1.set_xlabel("NLCD")

plt.savefig("../final_fig/era5_wu_scatter.pdf", dpi=1000, bbox_inches='tight')
```

```python
fig, (ax, ax1) = plt.subplots(2, 1, figsize=(2, 2))

x = np.linspace(0, 48, 100)
y = np.sin(2*np.pi*x/24)

ax.plot(x, y)
ax.set_xticks([])
ax.set_yticks([])

x = np.random.normal(0, 1, size=48)
x1 = np.cumsum(x)
ax1.set_yticks([])
ax1.set_xticks([])

ax1.plot(x1)
ax1.set_xlabel("Time (hours)")
ax1.set_ylabel("Random")
ax.set_ylabel("Cycle")

plt.savefig("../final_fig/cycle_random.pdf", dpi=1000, bbox_inches='tight')
```
