# Compare clusters
In this notebook, I walk through looking at the different clusters so that we can compare the temperatures between the two.

```python
import pandas as pd
import numpy as np
import osmnx as ox
import matplotlib.pyplot as plt

# Define your area of interest
place = "Durham, North Carolina, USA"  # or use a bounding box

west, south, east, north = -79.1, 35.8, -78.6, 36.3

G = ox.graph_from_bbox([west, south, east, north], network_type='drive', custom_filter='["highway"~"motorway|trunk"]')
edges = ox.graph_to_gdfs(G, nodes=False, edges=True)
```

```python
c0 = pd.read_csv("../experiments/clusters/cluster_0.csv", index_col=0)
lon_vals = c0.columns.values.astype(float)
lat_vals = c0.index.values
c0 = c0.values

plt.figure(figsize=(3, 4))
# cp = plt.contourf(lon_vals, lat_vals, value_grid, levels=3, cmap='Reds', vmax=2)

cp = plt.contourf(lon_vals, lat_vals, c0, levels=np.arange(0, 3.1, 0.25), cmap='Reds', vmax=3)
cbar = plt.colorbar(cp, label='°C', orientation='horizontal', shrink=0.7, pad=0.05)
# cbar.set_ticks([0, 0.5, 1, 1.5, 2])  # Explicitly set tick marks
edges.plot(ax=plt.gca(), color='black', linewidth=0.5, label='Major Roads')
plt.xlim(lon_vals.min(), lon_vals.max())
plt.ylim(lat_vals.min(), lat_vals.max())

plt.axis('off')

# plt.colorbar(cp, label='°C', orientation='horizontal', shrink=0.8, pad=0.05)
# plt.xlabel('Longitude')
# plt.ylabel('Latitude')
plt.title(f'Cluster {0}')

plt.savefig(f'../figures/cluster0.pdf', dpi=300, bbox_inches='tight')
plt.show()
```

```python
c3 = pd.read_csv("../experiments/clusters/cluster_3.csv", index_col=0)

lon_vals = c3.columns.values.astype(float)
lat_vals = c3.index.values
c3 = c3.values

plt.figure(figsize=(3, 4))
# cp = plt.contourf(lon_vals, lat_vals, value_grid, levels=3, cmap='Reds', vmax=2)

cp = plt.contourf(lon_vals, lat_vals, c3, levels=np.arange(0, 3.1, 0.25), cmap='Reds', vmax=3)
cbar = plt.colorbar(cp, label='°C', orientation='horizontal', shrink=0.7, pad=0.05)
# cbar.set_ticks([0, 0.5, 1, 1.5, 2])  # Explicitly set tick marks
edges.plot(ax=plt.gca(), color='black', linewidth=0.5, label='Major Roads')
plt.xlim(lon_vals.min(), lon_vals.max())
plt.ylim(lat_vals.min(), lat_vals.max())
plt.axis('off')

# plt.colorbar(cp, label='°C', orientation='horizontal', shrink=0.8, pad=0.05)
# plt.xlabel('Longitude')
# plt.ylabel('Latitude')
plt.title(f'Cluster {3}')
plt.savefig(f'../figures/cluster3.pdf', dpi=300, bbox_inches='tight')

plt.show()
```
